```python
# Setting up
import torch
print(torch.__version__)
print("GPU availableL:", torch.cuda.is_available())
```

```python
# Getting data
import urllib.request

url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"

urllib.request.urlretrieve(url, "input.txt")

with open("input.txt", "r", encoding="utf-8") as f:
  text = f.read()

print("length of dataset in characters:", len(text))
print(text[:300])
```

```python
# Tokenizers
chars = sorted(list(set(text)))

vocab_size = len(chars)

stoi = {ch:i for i, ch in enumerate(chars)}
itos = {i:ch for i, ch in enumerate(chars)}

encode = lambda s:[stoi[c] for c in s]
decode = lambda l: "".join([itos[i] for i in l])

print(encode("Hii there"))
print(decode(encode("Hii there")))
```

```python
# Training and Validation Sets
data = torch.tensor(encode(text), dtype=torch.long)
print(data.shape, data.dtype)
print(data[:100])

n = int(0.9 * len(data))
train_data = data[:n]
val_data = data[n:]
```

```python
# Chunks and Batches

# Define block size
block_size = 8

# Get the first block size chunk in train data
x = train_data[:block_size]

# Get the next shifted by one block size chunk in train data
# for testing
y = train_data[1:block_size+1]

# Loop through block size
for t in range(block_size):
  # Get context is everything up to and include t
  context = x[:t+1]
  # Get target which is at t since y is shifted by one
  target = y[t]
  # test
  print(f"when input is {context.tolist()} the target is {target}")
```
